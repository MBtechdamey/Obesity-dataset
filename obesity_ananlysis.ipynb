Executive Summary

Introduction

Obesity has become a major global health issue, contributing to the increasing rate of chronic diseases such as diabetes, hypertension, and cardiovascular problems. As lifestyle patterns continue to change, there is a growing need to understand the factors that influence obesity in order to support effective prevention and intervention strategies. This report presents an analysis of an obesity dataset to examine the demographic, behavioural, and lifestyle factors associated with different obesity classifications.






Data Overview

The analysis is based on a dataset comprising 2,111 observations and 17 variables. The dataset contains information on individuals’ demographic characteristics, including age and gender, as well as physical measurements such as height and weight. It also shows behavioural and lifestyle variables, including family history of overweight, dietary habits, frequency of physical activity, water consumption, smoking status, alcohol consumption, technology usage, transportation methods, and meal patterns. The target variable classifies individuals into different weight categories, ranging from underweight and normal weight to various levels of overweight and obesity, providing a comprehensive basis for analysing obesity-related trends.






Objectives

The primary objective of this study is to identify the factors that are most strongly associated with obesity levels. The analysis aims to:

Check the demographic and lifestyle characteristics of individuals in the dataset.
Explore relationships between behavioural factors and obesity classifications.
Identify key variables that contribute to increased obesity risk.
Generate insights that can support evidence 
 





 
Key Findings

The analysis shows that obesity is influenced by a combination of demographic, hereditary, and lifestyle related factors rather than a single determinant. Variables such as individual body weight, family history of overweight, physical activity, dietary habits, and eating patterns demonstrate strong associations with obesity classifications.Healthier lifestyle behaviours including regular physical activity, balanced eating habits, and adequate water intake are generally linked with lower obesity levels. These findings reinforce existing evidence that obesity is a serious condition that requires comprehensive prevention strategies that address both behavioural and genetic influences.
The largest obesity group was Obesity Type I, representing 16.8% of all participants, while 12.8% were classified as underweight. Most of the participants were between 20 and 25 years old, showing that the dataset mainly represents young adults.
The analysis also showed some important relationships between lifestyle factors and obesity. Nearly all participants in the obesity groups had a family history of overweight. Specifically, 98.0% of those in Obesity Type I, 99.7% in Obesity Type II, and 100% in Obesity Type III reported a family history of overweight.which clearly means its a genetic medical condition. Also, it was discovered in the dataset that individuals who consume alot of high calorie foods tends to have the condition.with over 96% of individuals in the obesity categories exhibiting this habit.

In addition, individuals with higher obesity levels generally showed lower levels of physical activity.which implies that, individuals who are physically active, have lesser chances of being obesed.

Overall, the findings suggest that obesity is influenced by a combination of family history and lifestyle habits rather than a single factor. Healthy eating, regular physical activity, and early awareness for individuals with a family history of overweight are important steps in reducing obesity risk. This project also shows how data analytics can be used to identify health trends and provide information that supports better healthcare decisions.







Recommendations

Based on the findings, greater emphasis should be placed on promoting healthy lifestyle behaviours through public health education and community based interventions. general awareness that encourages regular physical activity, balanced nutrition, and healthy eating habits should be prioritised, particularly for individuals with a family history of obesity. Healthcare providers should also consider lifestyle and hereditary factors when developing personalised prevention and weight management plans. Finally, medical practitioners should create systems that allows individuals monitore the basic things that helps them identify when they are at rik of obesity.

# Physical Activity vs. Obesity Level

```python

plt.figure(figsize=(10,6))
df.boxplot(column="FAF", by="NObeyesdad", grid=False)
plt.title("Physical Activity Frequency by Obesity Level")
plt.suptitle("")
plt.xlabel("Obesity Level")
plt.ylabel("Physical Activity Frequency (FAF)")
plt.xticks(rotation=45)
plt.show()
```

# This box plot shows the comparison between FAF against all the obesity levels

```python

# Create a cross-tabulation
family_obesity = pd.crosstab(
    df["NObeyesdad"],
    df["family_history_with_overweight"])
# Plot
family_obesity.plot(
    kind="bar",
    figsize=(10,6),
    color=["lightblue", "salmon"])

plt.title("Family History of Overweight by Obesity Level")
plt.xlabel("Obesity Level")
plt.ylabel("Number of Individuals")
plt.legend(title="Family History")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
